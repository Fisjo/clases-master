```python
import numpy as np
import pandas as pd

print(np.__version__)
print(pd.__version__)
```

```python
capacity = pd.Series([20, 15, 25, 30], index=["Xativa", "Colon", "Ruzafa", "Angel Gimera"])
print(capacity)
```

```python
print(capacity * 0.5)
```

```python
mini = pd.DataFrame({
    "Station": ["Xativa", "Colon", "Ruzafa", "Angel Gimera"],
    "Capacity": [20, 15, 25, 30],
    "Bikes": [12, 3, 18, 25]
})

print(mini)
```

```python
print(mini.columns.to_list())
print(mini.dtypes)
```

```python
mini["Ocupancy"] = (mini["Bikes"] / mini["Capacity"]).round(2)

mini["Status"] = np.where(mini["Ocupancy"] > 0.7, "High", "Okay")

print(mini)
```
