```python
mini = pd.DataFrame({
    "Station": ["Xativa", "Colon", "Ruzafa", "Angel Gimera"],
    "Capacity": [20, 15, 25, 30],
    "Bikes": [12, 3, 18, 25]
})

print(mini)
```

```python
mini["Ocupancy"] = (mini["Bikes"] / mini["Capacity"]).round(2)

mini["Status"] = np.where(mini["Ocupancy"] > 0.7, "High", "Okay")

print(mini)
```

```python
#reto 4

n_999 = (usage_hourly["rentals"] == 999).sum()
n_neg = (usage_hourly["returns"] < 0).sum()

print(f"Filas con 999 alquileres: {n_999}")
print(f"Filas con devoluciones negativas: {n_neg}")
```

```python
# reto 5

grandes = stations.loc[
    (stations["district"] == "Ciutat Vella") & (stations["capacity"] >= 30),
    "station_id"
].sort_values().tolist()

print(grandes)
```

```python
# reto 6

validas = usage_hourly[
    (usage_hourly["rentals"] != 999) & (usage_hourly["returns"] >= 0)
]

ranking = validas.sort_values(
    by=["returns", "timestamp", "station_id"],
    ascending=[False, True, True],
)

print(ranking.head(3))
print(ranking.iloc[0])
```

```python
# reto 7

clean = usage_hourly.copy()
n_inicial = len(clean)

clean = clean.drop_duplicates()
n_duplicados = n_inicial - len(clean)

n_antes = len(clean)
clean = clean[
    clean["rentals"].between(0, 100) & clean["returns"].between(0, 100)
]
n_imposibles = n_antes - len(clean)

print(f"Filas iniciales: {n_inicial}")
print(f"Duplicados quitados: {n_duplicados}")
print(f"Imposibles quitados: {n_imposibles}")
print(f"Total quitado: {n_inicial - len(clean)}")
print(f"Filas finales: {len(clean)}")





```

```python
# reto 8
clean["timestamp"] = pd.to_datetime(clean["timestamp"])

clean["date"] = clean["timestamp"].dt.date
clean["hour"] = clean["timestamp"].dt.hour
clean["weekday"] = clean["timestamp"].dt.day_name()
clean["is_weekend"] = clean["timestamp"].dt.dayofweek >= 5

print(clean[["timestamp", "date", "hour", "weekday", "is_weekend"]].head())
print(clean.dtypes)
```
