### 1. Importar NumPy

### 3. Crear un array 2D y ver sus atributos
Creamos una matriz de 3 filas × 4 columnas (por ejemplo, alquileres por franja) y consultamos:
- `shape`: forma (filas, columnas)
- `ndim`: número de dimensiones
- `size`: número total de elementos
- `dtype`: tipo de dato de los elementos

```python
rentals_grid = np.array([
    [2,8,5,1],
    [0,3,6,9],
    [6,6,6,0]
])

print(rentals_grid)

print(f"La forma es: {rentals_grid.shape}")
print(f"La dimensión es: {rentals_grid.ndim}")
print(f"El tamaño es: {rentals_grid.size}")
print(f"El tipo es: {rentals_grid.dtype}")
```

### 4. Conversión automática de tipos
NumPy exige que todos los elementos de un array sean del mismo tipo. Al mezclar números y texto, convierte todo a cadena (`str`), como muestra el `dtype`.

```python
print(np.array([1, "1.5", 3]).dtype)
```

### 5. Conversión de decimales a enteros
- `np.round(...).astype(int)`: redondea al entero más cercano y luego convierte.
- `astype(int)` directamente: **trunca** la parte decimal (no redondea).

```python
prices = np.array([1.9, 4.3, 5.2, 6.3])

print(np.round(prices).astype(int))

print(prices.astype(int))
```

### 6. Arrays de ceros y unos
`np.zeros` y `np.ones` crean arrays rellenos de 0 o 1 con la forma indicada. Útiles para inicializar datos.

```python
print(np.zeros(10))
print(np.ones([3,10]))
```

### 7. Secuencias numéricas
- `np.arange(inicio, fin, paso)`: valores con un paso fijo (el fin no se incluye).
- `np.linspace(inicio, fin, n)`: `n` valores equiespaciados entre inicio y fin (ambos incluidos).

```python
print(np.arange(0, 10, 5))
print(np.linspace(0, 10, 5))
```

### 8. Cambiar la forma con `reshape`
Creamos un array con las 24 horas del día y lo reorganizamos en distintas formas. Usar `-1` deja que NumPy calcule automáticamente esa dimensión.

```python
hours = np.arange(24)

print(hours)

hours = hours.reshape(4,6)
print(hours)

hours = hours.reshape(3, -1)
print(hours)

hours = hours.reshape(-1, 4)
print(hours)
```

### Indexado y slicing en un array 2D
Creamos una matriz de 4 × 5 con los números del 1 al 20 y accedemos a distintas partes:
- `grid[2, 4]`: un elemento (fila 2, columna 4)
- `grid[:, 3]`: toda la columna 3
- `grid[3, :]`: toda la fila 3

```python
grid = np.arange(1, 21).reshape(4,5)
print(grid)

print(grid[2,4])
print(grid[:, 3])
print(grid[3, :])
print(grid[::-1])
```

### Máscaras booleanas
Creamos un array de alquileres y la condición `rentals > 5`, que da un array de `True`/`False` (`busy`). Usar esa máscara como índice devuelve solo los valores que cumplen la condición.

```python
rentals = np.array([2, 3, 8, 5, 7, 3, 9])

busy = rentals > 5

print(busy)
print(rentals[busy])
```

### Contar y combinar condiciones
- `busy.sum()`: número de `True` (cuántos valores cumplen la condición)
- `busy.mean()`: proporción de `True`
- `(rentals >= 4) & (rentals <= 8)`: combina condiciones con `&`, cada una entre paréntesis, para filtrar valores entre 4 y 8

```python
print(busy.sum())
print(busy.mean())

print(rentals[(rentals >= 4) & (rentals <= 8)])

```

### Posiciones de los valores que cumplen una condición
- `np.argwhere(grid >= 17)`: lista de coordenadas `[fila, columna]` de cada valor que cumple la condición
- `np.where(grid > 17)`: las mismas posiciones, pero separadas en un array de filas y otro de columnas

```python
print(grid)
print(np.argwhere(grid >= 17))

print(np.where(grid > 17))
```

### Operaciones entre arrays
Las operaciones se aplican elemento a elemento. `rentals - returns` resta posición a posición, y `rentals * 1.5` multiplica cada valor por 1.5.

```python
rentals = np.array([5, 18, 2, 10])
returns = np.array([4, 9, 3, 6])

print(rentals - returns)
print(rentals * 1.5)
```

### Operaciones acumuladas y diferencias
Calculamos el balance neto (`returns - rentals`) y lo analizamos:
- `np.cumsum`: suma acumulada
- `np.cumprod`: producto acumulado
- `np.cumsum(net) + 12`: suma acumulada con un valor inicial de 12
- `np.diff`: diferencia entre elementos consecutivos

```python
net = returns - rentals

print(net)
print(np.cumsum(net))
print(np.cumprod(net))
print(np.cumsum(net) + 12)
print(np.diff(net))
```

### Reto 6 · Alerta sin romper nada
Creamos una copia de `bikes` con `.copy()` para no modificar el original, ponemos a 0 las lecturas menores que 5 con una máscara booleana y calculamos la media de la copia.

```python
alerta = bikes.copy()
alerta[alerta < 5] = 0

media_alerta = alerta.mean()
print(f"Media de la copia: {media_alerta}")
print(f"Media de bikes (original): {bikes.mean()}")
```

### Reto 7 · Condiciones combinadas
Seleccionamos las estaciones 0 y 2 con una lista de índices y contamos cuántas lecturas están entre 8 y 12 (ambos incluidos). Las condiciones se combinan con `&` y cada una va entre paréntesis; `.sum()` cuenta los `True`.

```python
seleccion = bikes[[0, 2], :]
en_rango = (seleccion >= 8) & (seleccion <= 12)
cantidad = en_rango.sum()

print(f"Lecturas entre 8 y 12 en las estaciones 0 y 2: {cantidad}")
```

**Comprobación del reto 7:** cuenta con un bucle las lecturas entre 8 y 12 de las estaciones 0 y 2 y comprueba que coincidan con `cantidad`.

```python
# check reto 7
assert seleccion.shape == (2, 24), "Debe haber 2 estaciones x 24 horas"
expected = sum(1 for e in (0, 2) for v in bikes[e] if 8 <= v <= 12)
assert cantidad == expected, "Revisa los límites (ambos incluidos) y las estaciones 0 y 2"
print("Reto 7 correcto")
```
