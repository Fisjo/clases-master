```python
rentals_grid = np.array([
    [2,8,5,1],
    [0,3,6,9],
    [6,6,6,0]
])

print(rentals_grid)

print(f"La forma es: {rentals_grid.shape}")
print(f"La dimensión es: {rentals_grid.ndim}")
print(f"El tamaño es: {rentals_grid.size}")
print(f"El tipo es: {rentals_grid.dtype}")
```

```python
print(np.array([1, "1.5", 3]).dtype)
```

```python
prices = np.array([1.9, 4.3, 5.2, 6.3])

print(np.round(prices).astype(int))

print(prices.astype(int))
```
