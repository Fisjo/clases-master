### 1. Importar NumPy
Importamos la librería `numpy` con el alias habitual `np`. Es la base para trabajar con arrays numéricos en data science.

```python
import numpy as np
```

### 3. Crear un array 2D y ver sus atributos
Creamos una matriz de 3 filas × 4 columnas (por ejemplo, alquileres por franja) y consultamos:
- `shape`: forma (filas, columnas)
- `ndim`: número de dimensiones
- `size`: número total de elementos
- `dtype`: tipo de dato de los elementos

```python
rentals_grid = np.array([
    [2,8,5,1],
    [0,3,6,9],
    [6,6,6,0]
])

print(rentals_grid)

print(f"La forma es: {rentals_grid.shape}")
print(f"La dimensión es: {rentals_grid.ndim}")
print(f"El tamaño es: {rentals_grid.size}")
print(f"El tipo es: {rentals_grid.dtype}")
```

### 4. Conversión automática de tipos
NumPy exige que todos los elementos de un array sean del mismo tipo. Al mezclar números y texto, convierte todo a cadena (`str`), como muestra el `dtype`.

```python
print(np.array([1, "1.5", 3]).dtype)
```

### 5. Conversión de decimales a enteros
- `np.round(...).astype(int)`: redondea al entero más cercano y luego convierte.
- `astype(int)` directamente: **trunca** la parte decimal (no redondea).

```python
prices = np.array([1.9, 4.3, 5.2, 6.3])

print(np.round(prices).astype(int))

print(prices.astype(int))
```

### 6. Arrays de ceros y unos
`np.zeros` y `np.ones` crean arrays rellenos de 0 o 1 con la forma indicada. Útiles para inicializar datos.

```python
print(np.zeros(10))
print(np.ones([3,10]))
```

### 7. Secuencias numéricas
- `np.arange(inicio, fin, paso)`: valores con un paso fijo (el fin no se incluye).
- `np.linspace(inicio, fin, n)`: `n` valores equiespaciados entre inicio y fin (ambos incluidos).

```python
print(np.arange(0, 10, 5))
print(np.linspace(0, 10, 5))
```

### 8. Cambiar la forma con `reshape`
Creamos un array con las 24 horas del día y lo reorganizamos en distintas formas. Usar `-1` deja que NumPy calcule automáticamente esa dimensión.

```python
hours = np.arange(24)

print(hours)

hours = hours.reshape(4,6)
print(hours)

hours = hours.reshape(3, -1)
print(hours)

hours = hours.reshape(-1, 4)
print(hours)
```

### 9. Números aleatorios enteros
Creamos un generador con semilla (`default_rng(42)`) para que los resultados sean reproducibles y generamos 20 enteros aleatorios entre 0 y 99.

```python
random_array = np.random.default_rng(42)

random_array.integers(0, 100, size=20)
```

### 10. Números aleatorios decimales

```python
random_array.random(10)
```
