```python
import numpy as np
```

```python
np.__version__
```

```python
rentals_grid = np.array([
    [2,8,5,1],
    [0,3,6,9],
    [6,6,6,0]
])

print(rentals_grid)

print(f"La forma es: {rentals_grid.shape}")
print(f"La dimensión es: {rentals_grid.ndim}")
print(f"El tamaño es: {rentals_grid.size}")
print(f"El tipo es: {rentals_grid.dtype}")
```

```python
print(np.array([1, "1.5", 3]).dtype)
```

```python
prices = np.array([1.9, 4.3, 5.2, 6.3])

print(np.round(prices).astype(int))

print(prices.astype(int))
```

```python
print(np.zeros(10))
print(np.ones([3,10]))
```

```python
print(np.arange(0, 10, 5))
print(np.linspace(0, 10, 5))
```

```python
hours = np.arange(24)

print(hours)

hours = hours.reshape(4,6)
print(hours)

hours = hours.reshape(3, -1)
print(hours)

hours = hours.reshape(-1, 4)
print(hours)
```

```python
random_array = np.random.default_rng(42)

random_array.integers(0, 100, size=20)
```

```python
random_array.random(10)
```

```python
random_array.poisson(lam=10, size=(5, 24))
```
